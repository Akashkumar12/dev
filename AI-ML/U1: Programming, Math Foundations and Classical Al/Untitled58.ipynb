```python
import numpy as np
from sklearn.linear_model import LinearRegression
X = np.array([[3,4],[4,5],[8,9],[5,5]])   #input
y = np.array([7,9,17,10])                 #output
```

```python
model = LinearRegression()
model.fit(X,y)
```

```python
print(round(model.predict([[1,2]])[0],0))
```

```python
model.predict([[10,10]])
```

```python
print(f"f(x) = {model.coef_[0]} *x1 + {model.coef_[1]} *x2")
```

```python
model.intercept_
```

```python
import joblib
```

```python
joblib.dump(model,"mymodel")
```

```python
model.score(X,y)
```
